# 21 — Phase 2: ASEAN & Asia-Pacific Deep-Dive (`v2.1.0`)

**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Inspect the 17-jurisdiction ASEAN/APAC matrix (11 covered, 6 identified gaps) and plot the 500-program SEA portfolio.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
mat = pd.read_csv(ROOT / 'data' / 'curated' / 'asean_apac_comparative_matrix.csv')
sea = pd.read_parquet(ROOT / 'data' / 'curated' / 'asean_sea_programs_500.parquet')
print(mat['coverage_status'].value_counts())
print('SEA 500 programs mean quality:', round(sea['quality_score'].mean(), 2))

fig, ax = plt.subplots(figsize=(8, 4))
sea['human_rights_risk_tier'].value_counts().plot(kind='barh', ax=ax, color='#1e40af')
ax.set_title('500 SEA Research Programs by Human-Rights Risk Tier')
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'asean_sea_500_summary.png', dpi=150)
print('Saved chart to reports/asean_sea_500_summary.png')
```
