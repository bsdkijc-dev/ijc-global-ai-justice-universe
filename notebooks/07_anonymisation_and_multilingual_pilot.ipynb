# 07 — Offline Judgment Anonymisation & Multilingual Justice Glossary

**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Verify `anonymisation_gate.py` (`PROTOTYPE_EXPERIMENTAL_NONBINDING`) and inspect the 50-term 5-language glossary.

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
anon = json.loads((ROOT / 'data' / 'sample' / 'anonymisation_benchmark_results.json').read_text(encoding='utf-8'))
df_anon = pd.DataFrame(anon['results'])
glossary = pd.read_csv(ROOT / 'data' / 'curated' / 'multilingual_glossary_50.csv')
print('Anonymisation accuracy:', anon['exact_redaction_accuracy'], '| Glossary terms:', len(glossary))

fig, ax = plt.subplots(figsize=(7, 3.8))
x = range(len(df_anon))
ax.bar([i - 0.15 for i in x], df_anon['expected_redactions'], width=0.3, label='Expected', color='#1e40af')
ax.bar([i + 0.15 for i in x], df_anon['detected_redactions'], width=0.3, label='Detected', color='#2f855a')
ax.set_xticks(list(x))
ax.set_xticklabels(df_anon['case_id'])
ax.set_title('Offline Anonymisation Gate Benchmark (Synthetic Suite)')
ax.legend()
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'anonymisation_benchmark_summary.png', dpi=150)
print('Saved chart to reports/anonymisation_benchmark_summary.png')
```
