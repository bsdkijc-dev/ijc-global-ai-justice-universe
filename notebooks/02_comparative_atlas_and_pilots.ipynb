**Purpose:** Comparative analysis of 46 Oxford AI Justice Atlas jurisdictions and the 1,000-program curated portfolio.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
atlas = pd.read_csv(ROOT / 'data' / 'curated' / 'jurisdiction_atlas_46.csv')
curated = pd.read_parquet(ROOT / 'data' / 'curated' / 'programs_curated_1000.parquet')
pilots = pd.read_csv(ROOT / 'data' / 'curated' / 'top10_pilot_prototypes.csv')

print('Jurisdictions:', len(atlas), '| Accessible:', (atlas['verification_status'] == 'VERIFIED_ACCESSIBLE').sum())
print('Curated programs:', len(curated), '| Mean quality:', round(curated['quality_score'].mean(), 2))
print('Top-10 pilots regions:', pilots['D01_jurisdiction_archetype'].nunique())
```

```python
acc = atlas[atlas['verification_status'] == 'VERIFIED_ACCESSIBLE']
indicators = {
    'Judicial AI Policy': int(acc['has_judicial_ai_policy'].sum()),
    'AI Law / Strategy': int(acc['has_national_ai_law_or_strategy'].sum()),
    'FRT / Biometric Pushback': int(acc['has_biometric_or_frt_pushback'].sum()),
    'Hallucination Cases': int(acc['has_hallucination_cases'].sum()),
    'Multilingual Court ASR': int(acc['has_multilingual_court_asr'].sum()),
}
fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))
pd.Series(indicators).plot(kind='barh', ax=axes[0], color='#2b6cb0')
axes[0].set_title('Comparative Justice AI Indicators (n=45 Accessible)')
curated['D01_jurisdiction_archetype'].value_counts().plot(kind='bar', ax=axes[1], color='#2f855a')
axes[1].set_title('Curated 1,000 Programs by Region Archetype')
axes[1].tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'comparative_atlas_summary.png', dpi=150)
print('Saved chart to reports/comparative_atlas_summary.png')
```
