# 24 — Phase 2: Sub-Saharan Africa & MENA Deep-Dive (`v2.4.0`)

**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Inspect the 14-jurisdiction SSA/MENA matrix (13 accessible, Zambia `SOURCE_INACCESSIBLE`) and plot the 500-program portfolio.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
mat = pd.read_csv(ROOT / 'data' / 'curated' / 'ssa_mena_comparative_matrix.csv')
df = pd.read_parquet(ROOT / 'data' / 'curated' / 'ssa_mena_programs_500.parquet')
print(mat['verification_status'].value_counts())
print('SSA/MENA 500 mean quality:', round(df['quality_score'].mean(), 2))

fig, ax = plt.subplots(figsize=(8, 4))
df['human_rights_risk_tier'].value_counts().plot(kind='barh', ax=ax, color='#1e40af')
ax.set_title('500 SSA & MENA Research Programs by Human-Rights Risk Tier')
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'ssa_mena_500_summary.png', dpi=150)
print('Saved chart to reports/ssa_mena_500_summary.png')
```
