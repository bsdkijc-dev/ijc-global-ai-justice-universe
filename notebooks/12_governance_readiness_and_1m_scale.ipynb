# 12 — Judicial AI Governance Readiness Index & 1M Scale Benchmark

**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Plot top jurisdictions by Governance Readiness Score and inspect the 1,000,000-program streaming benchmark.

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
df = pd.read_csv(ROOT / 'data' / 'curated' / 'governance_readiness_46.csv').dropna(subset=['readiness_score'])
m = json.loads((ROOT / 'data' / 'manifests' / 'scale_1m_manifest.json').read_text(encoding='utf-8'))
print('Scored countries:', len(df), '| Mean readiness:', round(df['readiness_score'].mean(), 2))
print('1M Benchmark:', m['record_count'], 'records in', m['generation_seconds'], 's')

top15 = df.head(15).iloc[::-1]
fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(top15['jurisdiction_name'], top15['readiness_score'], color='#1e40af')
ax.set_title('Top 15 Jurisdictions — Judicial AI Governance Activity Score (n=45 Accessible)')
ax.set_xlabel('Score (0-100)')
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'governance_readiness_chart.png', dpi=150)
print('Saved chart to reports/governance_readiness_chart.png')
```
