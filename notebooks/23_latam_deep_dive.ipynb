# 23 — Phase 2: Latin America Constitutional & Judicial AI Deep-Dive (`v2.3.0`)

**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Inspect the 5-country LATAM matrix and plot the 500-program LAT portfolio.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
mat = pd.read_csv(ROOT / 'data' / 'curated' / 'latam_comparative_matrix.csv')
lat = pd.read_parquet(ROOT / 'data' / 'curated' / 'latam_lat_programs_500.parquet')
print('LATAM jurisdictions:', len(mat), '| LAT 500 mean quality:', round(lat['quality_score'].mean(), 2))

fig, ax = plt.subplots(figsize=(8, 4))
lat['human_rights_risk_tier'].value_counts().plot(kind='barh', ax=ax, color='#1e40af')
ax.set_title('500 LAT Research Programs by Human-Rights Risk Tier')
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'latam_lat_500_summary.png', dpi=150)
print('Saved chart to reports/latam_lat_500_summary.png')
```
