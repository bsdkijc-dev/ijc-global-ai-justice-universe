**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Inspect the 7-jurisdiction NAM/ECA matrix and plot the 500-program NAM/ECA portfolio.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
mat = pd.read_csv(ROOT / 'data' / 'curated' / 'nam_eca_comparative_matrix.csv')
df = pd.read_parquet(ROOT / 'data' / 'curated' / 'nam_eca_programs_500.parquet')
print('NAM/ECA jurisdictions:', len(mat), '| Mean quality:', round(df['quality_score'].mean(), 2))

fig, ax = plt.subplots(figsize=(8, 4))
df['human_rights_risk_tier'].value_counts().plot(kind='barh', ax=ax, color='#1e40af')
ax.set_title('500 NAM & ECA Research Programs by Human-Rights Risk Tier')
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'nam_eca_500_summary.png', dpi=150)
print('Saved chart to reports/nam_eca_500_summary.png')
```
