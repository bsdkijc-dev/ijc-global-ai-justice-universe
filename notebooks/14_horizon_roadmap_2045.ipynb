# 14 — Long-Horizon Research Roadmap (2027–2045)

**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Plot the 2027–2045 horizon distribution across 100,000 curated programs and verify the unified query CLI.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
df = pd.read_csv(ROOT / 'data' / 'curated' / 'horizon_roadmap_2027_2045.csv')
print('Horizons:', len(df), '| Total programs:', int(df['programs_100k_count'].sum()))

fig, ax = plt.subplots(figsize=(9, 4.2))
ax.bar(df['horizon_code'], df['programs_100k_count'], color='#1e40af')
ax.set_title('100,000 Curated Programs Across 2027-2045 Horizons')
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'horizon_roadmap_2045.png', dpi=150)
print('Saved chart to reports/horizon_roadmap_2045.png')
```
