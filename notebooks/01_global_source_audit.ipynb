**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)

**Purpose:** Reproducible audit of the 47 Oxford AI Justice Atlas URLs and the 100 deterministic candidate programs.

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../sources').exists() else Path('.').resolve()
reg = pd.read_csv(ROOT / 'sources' / 'source-registry.csv')
print('Total sources:', len(reg))
print(reg['verification_status'].value_counts())
```

```python
prog_df = pd.read_parquet(ROOT / 'data' / 'sample' / 'programs_sample_100.parquet')
print('Sample programs:', len(prog_df))
print('Mean quality score:', round(prog_df['quality_score'].mean(), 2))
print('Mean originality score:', round(prog_df['originality_score'].mean(), 4))
print(prog_df['human_rights_risk_tier'].value_counts())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
reg['verification_status'].value_counts().plot(kind='bar', ax=axes[0], color=['#2b6cb0', '#c53030'])
axes[0].set_title('Oxford AI Atlas Source Status (n=47)')
axes[0].tick_params(axis='x', rotation=15)
prog_df['human_rights_risk_tier'].value_counts().plot(kind='bar', ax=axes[1], color='#2f855a')
axes[1].set_title('Sampled Programs by Risk Tier (n=100)')
axes[1].tick_params(axis='x', rotation=25)
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'source_audit_summary.png', dpi=150)
print('Saved chart to reports/source_audit_summary.png')
```
