# 06 — 10,000-Program Scale-Up & Policy Analytics

**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Analyze the 10,000 validated 12D programs (`seed=202706`) across risk tiers and regional archetypes.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
df = pd.read_parquet(ROOT / 'data' / 'curated' / 'programs_curated_10000.parquet')
print('Total programs:', len(df), '| Unique hashes:', df['tuple_sha256'].nunique())
print('Mean quality:', round(df['quality_score'].mean(), 2))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))
df['human_rights_risk_tier'].value_counts().plot(kind='barh', ax=axes[0], color='#1e40af')
axes[0].set_title('10,000 Programs by Human-Rights Risk Tier')
df['D01_jurisdiction_archetype'].value_counts().plot(kind='bar', ax=axes[1], color='#2f855a')
axes[1].set_title('10,000 Programs by Regional Archetype')
axes[1].tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'scale_10k_distribution.png', dpi=150)
print('Saved chart to reports/scale_10k_distribution.png')
```
