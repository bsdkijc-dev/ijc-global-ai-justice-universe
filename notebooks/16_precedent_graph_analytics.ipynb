# 16 — Cross-Jurisdictional Precedent & Governance Graph

**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Plot theme degree centrality across 45 verified accessible jurisdictions.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
df = pd.read_csv(ROOT / 'data' / 'curated' / 'precedent_graph_edges.csv')
print('Verified graph edges:', len(df), '| Jurisdictions:', df['jurisdiction'].nunique())

fig, ax = plt.subplots(figsize=(8.5, 4.2))
df['target_label'].value_counts().iloc[::-1].plot(kind='barh', ax=ax, color='#1e40af')
ax.set_title('Precedent & Governance Theme Degree Centrality (n=45 Accessible)')
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'precedent_graph_summary.png', dpi=150)
print('Saved chart to reports/precedent_graph_summary.png')
```
