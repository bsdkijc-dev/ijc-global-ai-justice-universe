# 09 — Risk-Audit Disparity Simulator & 30-Pillar Coverage

**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Verify synthetic risk-scoring disparity detection and plot 30-pillar distribution across 10,000 programs.

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
risk = json.loads((ROOT / 'data' / 'sample' / 'risk_audit_benchmark_results.json').read_text(encoding='utf-8'))
df = pd.read_csv(ROOT / 'data' / 'curated' / 'pillar_coverage_30.csv')
print('Risk Audit Status:', risk['audit_status'], '| FPR gap:', risk['false_positive_rate_gap'])

fig, ax = plt.subplots(figsize=(11, 5))
ax.bar(df['pillar_id'], df['programs_10k_count'], color='#1e40af')
ax.set_title('10,000 Curated Programs Across 30 Global Research Pillars')
ax.tick_params(axis='x', rotation=60)
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'pillar_coverage_summary.png', dpi=150)
print('Saved chart to reports/pillar_coverage_summary.png')
```
