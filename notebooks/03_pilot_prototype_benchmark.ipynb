# 03 — Pilot Prototype Benchmark & Release Bundle Verification

**Project:** IJC Global AI & Justice Innovation Universe (2027–2045)  
**Purpose:** Verify the offline Citation Verification Gate & Appellate Triage Prototype (`PROTOTYPE_EXPERIMENTAL_NONBINDING`) and v0.3.0 release bundle.

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve() if Path('../data').exists() else Path('.').resolve()
bench = json.loads((ROOT / 'data' / 'sample' / 'prototype_benchmark_results.json').read_text(encoding='utf-8'))
df = pd.DataFrame(bench['results'])
print('Prototype accuracy:', bench['exact_citation_detection_accuracy'])
print(df[['case_id', 'cluster_id', 'citations_detected', 'expected_unverified', 'detected_unverified', 'gate_status']])
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
x = range(len(df))
ax.bar([i - 0.15 for i in x], df['expected_unverified'], width=0.3, label='Expected Unverified', color='#2b6cb0')
ax.bar([i + 0.15 for i in x], df['detected_unverified'], width=0.3, label='Detected Unverified', color='#2f855a')
ax.set_xticks(list(x))
ax.set_xticklabels(df['case_id'])
ax.set_title('Pilot Prototype Citation Gate Benchmark (Synthetic Suite)')
ax.legend()
plt.tight_layout()
plt.savefig(ROOT / 'reports' / 'prototype_benchmark_summary.png', dpi=150)
print('Saved chart to reports/prototype_benchmark_summary.png')
```
